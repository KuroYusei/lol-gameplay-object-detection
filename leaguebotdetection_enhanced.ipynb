## 📊 Evaluation & Model Metrics

```python
# ─────────────────────────────────────────────
# 1. TRAINING CURVES
# ─────────────────────────────────────────────
import os, pandas as pd, matplotlib.pyplot as plt, matplotlib.gridspec as gridspec

results_csv = "EsportsAnalyzer/YOLOv8m/results.csv"

if os.path.exists(results_csv):
    df = pd.read_csv(results_csv)
    df.columns = df.columns.str.strip()          # YOLO sometimes adds spaces

    fig = plt.figure(figsize=(18, 10))
    gs  = gridspec.GridSpec(2, 3, figure=fig, hspace=0.4, wspace=0.35)

    def safe_plot(ax, col, title, ylabel, color):
        if col in df.columns:
            ax.plot(df[col], color=color, linewidth=2)
            ax.set_title(title, fontsize=12, fontweight="bold")
            ax.set_xlabel("Epoch")
            ax.set_ylabel(ylabel)
            ax.grid(True, alpha=0.3)
        else:
            ax.text(0.5, 0.5, f"{col}\nnot found", ha="center", va="center",
                    transform=ax.transAxes, color="gray")
            ax.set_title(title, fontsize=12)

    safe_plot(fig.add_subplot(gs[0,0]), "train/box_loss",   "Train Box Loss",      "Loss",  "#e74c3c")
    safe_plot(fig.add_subplot(gs[0,1]), "train/cls_loss",   "Train Class Loss",    "Loss",  "#e67e22")
    safe_plot(fig.add_subplot(gs[0,2]), "train/dfl_loss",   "Train DFL Loss",      "Loss",  "#9b59b6")
    safe_plot(fig.add_subplot(gs[1,0]), "val/box_loss",     "Val Box Loss",        "Loss",  "#3498db")
    safe_plot(fig.add_subplot(gs[1,1]), "metrics/mAP50(B)", "mAP@0.50",            "mAP",   "#27ae60")
    safe_plot(fig.add_subplot(gs[1,2]), "metrics/mAP50-95(B)", "mAP@0.50-0.95",   "mAP",   "#1abc9c")

    fig.suptitle("YOLOv8m — Training Curves", fontsize=16, fontweight="bold", y=1.01)
    plt.savefig("training_curves.png", bbox_inches="tight", dpi=150)
    plt.show()
    print("✅ Training curves saved → training_curves.png")
else:
    print(f"⚠️  results.csv not found at: {results_csv}")
    print("    Run model.train() first, then re-run this cell.")
```

```python
# ─────────────────────────────────────────────
# 2. PRECISION / RECALL / F1 CURVES
# ─────────────────────────────────────────────
import os, pandas as pd, matplotlib.pyplot as plt

results_csv = "EsportsAnalyzer/YOLOv8m/results.csv"

if os.path.exists(results_csv):
    df = pd.read_csv(results_csv)
    df.columns = df.columns.str.strip()

    # Compute F1 from precision + recall if available
    p_col = "metrics/precision(B)"
    r_col = "metrics/recall(B)"

    fig, axes = plt.subplots(1, 3, figsize=(16, 4))

    for ax, col, title, color in zip(
        axes,
        [p_col, r_col, None],
        ["Precision", "Recall", "F1-Score"],
        ["#2ecc71", "#e74c3c", "#f39c12"]
    ):
        if title == "F1-Score":
            if p_col in df.columns and r_col in df.columns:
                p = df[p_col]; r = df[r_col]
                f1 = 2 * p * r / (p + r + 1e-9)
                ax.plot(f1, color=color, linewidth=2)
            else:
                ax.text(0.5, 0.5, "Precision/Recall\nnot found", ha="center",
                        va="center", transform=ax.transAxes, color="gray")
        else:
            if col in df.columns:
                ax.plot(df[col], color=color, linewidth=2)
            else:
                ax.text(0.5, 0.5, f"{col}\nnot found", ha="center",
                        va="center", transform=ax.transAxes, color="gray")
        ax.set_title(title, fontsize=12, fontweight="bold")
        ax.set_xlabel("Epoch")
        ax.grid(True, alpha=0.3)

    fig.suptitle("Precision / Recall / F1 over Training", fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig("pr_f1_curves.png", dpi=150)
    plt.show()
    print("✅ PR / F1 curves saved → pr_f1_curves.png")
else:
    print("⚠️  results.csv not found — run training first.")
```

## Overlay

#### captures your screen, runs detections, and shows a small always-on-top window with bounding boxes. It uses a Windows trick (SetWindowDisplayAffinity) to hide the overlay from screen recording software

```python
import cv2
import numpy as np
from mss import mss
import torch
from ultralytics import YOLO
import ctypes # <-- Added to talk to Windows directly

def main():
    device = "0" if torch.cuda.is_available() else "cpu"
    
    # 1. Path to your custom trained weights
    model_path = "best (1).pt"
    model = YOLO(model_path)

    # 2. Initialize screen capture
    sct = mss()
    monitor = sct.monitors[1] 

    # 3. Create the Always-On-Top Window
    window_name = "YOLO Overlay"
    cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
    cv2.setWindowProperty(window_name, cv2.WND_PROP_TOPMOST, 1)
    cv2.resizeWindow(window_name, 400, 225) 
    
    # Force a quick UI update so the window physically exists before we search for it
    cv2.waitKey(1) 

    # =================================================================
    # MAGIC TRICK: HIDE THE WINDOW FROM SCREEN RECORDERS (AND MSS)
    # =================================================================
    
    # Find the internal Windows handle (HWND) for our specific window
    hwnd = ctypes.windll.user32.FindWindowW(None, window_name)
    
    if hwnd:
        # WDA_EXCLUDEFROMCAPTURE = 0x11
        # This tells Windows 10/11: "Let the human see this window, 
        # but completely hide it from any screen capture software!"
        ctypes.windll.user32.SetWindowDisplayAffinity(hwnd, 0x11)
        print("Overlay successfully cloaked from capture.")
    else:
        print("Warning: Could not find window to cloak.")
        
    # =================================================================

    print("Overlay active. Drag it to a corner, click your game, and play!")

    while True:
        # 4. Capture screen (Because of the trick above, mss will see right through the overlay!)
        screenshot = sct.grab(monitor)
        frame = np.array(screenshot)[:, :, :3] 

        # 5. Run inference
        results = model.predict(source=frame, device=device, conf=0.30, verbose=False)
        annotated_frame = results[0].plot()

        # 6. Display the live feed in the pinned window
        cv2.imshow(window_name, annotated_frame)

        # Press 'q' inside the overlay window to close it
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
```

## detection by monitor

### Opens a full window showing what the model sees live.

```python
import cv2
import numpy as np
from mss import mss
import torch
from ultralytics import YOLO

def run_live_monitor_capture():
    # 1. Verify GPU acceleration is active
    device = "0" if torch.cuda.is_available() else "cpu"
    print(f"Using device: {torch.cuda.get_device_name(0) if device == '0' else 'CPU'}")

    # 2. Load your custom trained model
    model_path = "best (1).pt"
    
    print("Loading YOLOv8 model...")
    model = YOLO(model_path)
    print("Model loaded successfully.")

    # 3. Initialize screen capture (mss is incredibly fast for live feeds)
    sct = mss()
    
    # sct.monitors[1] targets your primary monitor
    # (Change to 2 if you are playing on a second screen)
    monitor = sct.monitors[1] 

    print("\n====================================================")
    print("LIVE MONITOR DETECTION ACTIVE")
    print("  -> Click on the 'Live Esports Analyzer' window")
    print("  -> Press the 'q' key on your keyboard to stop")
    print("====================================================\n")

    while True:
        # 4. Capture the screen
        screenshot = sct.grab(monitor)
        
        # 5. Convert raw BGRA pixels to a standard 3-channel numpy array (BGR)
        frame = np.array(screenshot)[:, :, :3] 

        # 6. Run inference using your RTX 3050 Ti
        # conf=0.30 filters out weak predictions so the AI is more accurate
        results = model.predict(source=frame, device=device, conf=0.30, verbose=False)
        
        # 7. Extract the frame with bounding boxes plotted
        annotated_frame = results[0].plot()

        # 8. Display the live visualization window
        cv2.imshow("Live Esports Analyzer", annotated_frame)

        # 9. Break the loop immediately if 'q' is pressed
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("Stopping live detection stream...")
            break

    # Clean up and close all windows safely
    cv2.destroyAllWindows()

# Run the function in this cell
run_live_monitor_capture()
```
